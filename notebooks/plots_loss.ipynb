YOLOv8n plots:

```python
import pandas as pd
import matplotlib.pyplot as plt

# путь к твоему results.csv
csv_path = "/Users/zahar/PycharmProjects/RDD_CV/YOLOv8n/project/results.csv"

df = pd.read_csv(csv_path)

# на всякий случай убираем пробелы в названиях колонок
df.columns = df.columns.str.strip()

# ----------------------------
# 1. mAP графики
# ----------------------------
plt.figure(figsize=(10,5))
plt.plot(df["epoch"], df["metrics/mAP50(B)"], label="mAP@0.5")
plt.plot(df["epoch"], df["metrics/mAP50-95(B)"], label="mAP@0.5:0.95")
plt.title("Метрика mAP по эпохам")
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.grid()
plt.legend()
plt.show()

# ----------------------------
# 2. Precision / Recall
# ----------------------------
plt.figure(figsize=(10,5))
plt.plot(df["epoch"], df["metrics/precision(B)"], label="Precision")
plt.plot(df["epoch"], df["metrics/recall(B)"], label="Recall")
plt.title("Precision и Recall по эпохам")
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.grid()
plt.legend()
plt.show()

# ----------------------------
# 3. Loss (обучение)
# ----------------------------
plt.figure(figsize=(10,5))
plt.plot(df["epoch"], df["train/box_loss"], label="train box loss")
plt.plot(df["epoch"], df["train/cls_loss"], label="train cls loss")
plt.plot(df["epoch"], df["train/dfl_loss"], label="train dfl loss")
plt.title("Train Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid()
plt.legend()
plt.show()

# ----------------------------
# 4. Loss (валидация)
# ----------------------------
plt.figure(figsize=(10,5))
plt.plot(df["epoch"], df["val/box_loss"], label="val box loss")
plt.plot(df["epoch"], df["val/cls_loss"], label="val cls loss")
plt.plot(df["epoch"], df["val/dfl_loss"], label="val dfl loss")
plt.title("Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid()
plt.legend()
plt.show()
```
